```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document

doc1 = Document(
    page_content="Babar Azam is pakistan's No 1 player.",
    metadata= {"team":"Peshawar Zalmi"}
)

doc2 = Document(
    page_content="Shaheen Shah Afridi is pakistan's pemiuim bowler.",
    metadata= {"team":"Islamabad United"}
)

doc3 = Document(
    page_content="Fakhar zaman is backbone of pakistan. He is a left hand batsman.",
    metadata= {"team":"Lahore Qalandars"}
)

doc4 = Document(
    page_content="Muhammad Hafeez is known as professor due to his unique batting techniques and he is retired now.He was an All rounder",
    metadata= {"team":"Lahore Qalandars"}
)

doc5 = Document(
    page_content="Haris Rouf is fast blower known for his pace in bowling.",
    metadata= {"team":"Lahore Qalandars"}
)

```

```python
docs = [doc1,doc2,doc3,doc4,doc5]
```

```python
print(docs)
```

```python
vector_store = Chroma(
   embedding_function=GoogleGenerativeAIEmbeddings(model="gemini-embedding-001"),
    persist_directory='chroma_db',
    collection_name='sample'
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings','documents','metadatas'])
```

## Similarity Search

```python
vector_store.similarity_search(
    query="who among these a bowler?",
    k=2
)
```

## Similarity Search With Score

```python
vector_store.similarity_search_with_score(
    query="who among these a fast bowler?",
    k=2
)
```

## Meta data filtering

```python
vector_store.similarity_search_with_score(
    query=" ",
    filter={"team":"Lahore Qalandars"}
)
```

## update Document

```python
updated_doc1 = Document(
    page_content="Babar Azam is pakistan's No 1 player.But now a days he is struggling with trolling",
    metadata= {"team":"Multan Sultan"}
)

vector_store.update_document(document_id=
  'f8e8cdde-1065-4313-ae3a-33d03bb3af2f', document=updated_doc1)
```

```python
vector_store.get(include=['embeddings','documents','metadatas'])
```
