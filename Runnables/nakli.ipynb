```python
from abc import ABC,abstractmethod
```

```python
class Runnable(ABC):
    @abstractmethod
    def invoke(input_data):
        pass
```

## LLM

```python
import random
class Naklillm(Runnable):
    def __init__(self):
        print('LLM created')
    def invoke(self,prompt):
        response_list = [
                'Islamabad is the capital of Pakistan',
                'PSL is the league played in pakistan',
                'AI stands for Artificiall intelligence'
                ]
        return {'response':random.choice(response_list)}
        
    def predict(self,prompt):
        response_list = [
        'Islamabad is the capital of Pakistan',
        'PSL is the league played in pakistan',
        'AI stands for Artificiall intelligence'
        ]
        return {'response':random.choice(response_list)}
        
```

```python
llm = Naklillm()
```

```python
llm.predict('What is the capital of Pakistan')
```

## Prompt Template

```python
class NakliPromptTemplate(Runnable):
    def __init__(self,template,input_variables):
        self.template=template
        self.input_variables= input_variables
    def invoke(self,input_dict):
        return self.template.format(**input_dict)
    def format(self,input_dict):
        return self.template.format(**input_dict)
    
```

```python
template = NakliPromptTemplate(
    template='Write a {lenght} poem about {topic}',
    input_variables=['lenght','topic']
)
```

```python
prompt = template.format({'lenght':'short','topic':'Pakistan'})
```

 ## Connecting prompt with LLM

```python
llm.predict(prompt)
```

## Nakli LLM Chain

```python
class NaklillmChain:
    def __init__(self,llm,prompt):
        self.llm=llm
        self.prompt=prompt
    def run(self,input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)
        return result['response']
```

```python
chain = NaklillmChain(llm,template)
chain.run({'lenght':'long','topic':'pakistan'})
```

```python
class Runnableconnector(Runnable):
    def __init__(self,runnable_list):
        self.runnable_list=runnable_list
        
    def  invoke(self,input_data):
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
            return input_data
```

```python
template = NakliPromptTemplate(
    template='Write a {lenght} poem about {topic}',
    input_variables=['lenght','topic']
)
```

```python
llm=Naklillm()
```

```python
chain = Runnableconnector([template,llm])
```

```python
chain.invoke({'lenght':'long','topic':'Pakistan'})
```
